# 10. List Comprehension & Exercises

```python
# Filtering list elements using a standard for loop
data = [10, 20, 30, 40, 50]
for i in data:
    if i >= 30:
        print(i)
```

```python
# Squaring elements using a standard for loop
data = [1, 2, 3, 4, 5]
empty = []
for i in data:
    x = i ** 2
    empty.append(x)
print(empty)
```

```python
# Squaring elements using List Comprehension
data = [1, 2, 3, 4, 5]
empty = [x ** 2 for x in data]
print(empty)
```

```python
# Filtering even numbers using List Comprehension
data = [23, 17, 44, 66, 21, 8]
even = [x for x in data if x % 2 == 0]
print(even)
```

```python
# Pass/Fail classification using List Comprehension (if-else)
data = [76, 45, 80, 34, 23, 85]
result = ["Pass" if x >= 60 else "Fail" for x in data]
print(result)
```

```python
# Converting strings to uppercase using List Comprehension
names = ["Amit", "Priya", "Rahul", "Soniya", "Sumit", "John"]
result = [n.upper() for n in names]
print(result)
```

```python
# Cleaning whitespace from strings using List Comprehension
names = [" Amit ", "   Priya ", " Rahul", "  Soniya ", "Sumit  ", "John"]
cleand_Data = [x.strip() for x in names]
print(names)
print(cleand_Data)
```

```python
# Extracting values from a list of dictionaries using List Comprehension
users = [
    {"name": "Rahul", "age": 25},
    {"name": "Priya", "age": 22}
]

# [Rahul, Priya]
empNames = [x["name"] for x in users]
print(empNames)
```

## Solve All These Problems Using List Comprehension (Q1 to Q18)

```python
# Q1. Find numbers greater than 50 and store all the numbers into new list.
numbers = [20, 55, 30, 80, 45, 90, 10]
result = [ex for ex in numbers if ex > 50]
print(result)
```

```python
# Q2. Find numbers between 20 and 50 and store all the numbers into new list.
numbers = [10, 25, 35, 60, 45, 15, 50, 70]
result = [ex for ex in numbers if 20 <= ex <= 50]
print(result)
```

```python
# Q3. Find numbers divisible by both 3 and 5 and store all the numbers into new list.
numbers = [10, 15, 20, 30, 45, 50, 60, 75]
result = [ex for ex in numbers if ex % 3 == 0 and ex % 5 == 0]
print(result)
```

```python
# Q4. Find names having more than 4 characters and store result into new list.
names = ["Amit", "Rahul", "Neha", "Rohit", "Sneha"]
result = [name for name in names if len(name) > 4]
print(result)
```

```python
# Q5. Find names starting with "A" and store result into new list.
names = ["Amit", "Rahul", "Ankit", "Neha", "Arjun", "Rohit"]
result = [name for name in names if name.startswith("A")]
print(result)
```

```python
# Q6. Extract vowels from a string and store result into new list.
text = "programming"
vowels = [ch for ch in text if ch.lower() in "aeiou"]
print(vowels)
```

```python
# Q7. Extract digits from a string and store result into new list.
data = "Python123Programming456"
digits = [ch for ch in data if ch.isdigit()]
print(digits)
```

```python
# Q8. Replace every even number with "Even" and every odd number with "Odd" and store result into new list.
numbers = [10, 15, 20, 25, 30]
result = ["Even" if ex % 2 == 0 else "Odd" for ex in numbers]
print(result)
```

```python
# Q9. Replace negative numbers with 0 and store result into new list.
numbers = [10, -5, 20, -8, 30, -2]
result = [ex if ex > 0 else 0 for ex in numbers]
print(result)
```

```python
# Q10. Give a 10% discount to products costing Rs. 1000 or more, otherwise give 5% and store result into new list.
prices = [500, 1200, 800, 2000, 1500]
result = [ex * 0.90 if ex >= 1000 else ex * 0.95 for ex in prices]
print(result)
```

```python
# Q12. Employee Management: Find employees earning more than Rs. 50,000.
employees = [
    {"name": "Amit", "salary": 45000},
    {"name": "Rahul", "salary": 60000},
    {"name": "Neha", "salary": 75000},
    {"name": "Rohit", "salary": 35000}
]

result = [ex['name'] for ex in employees if ex['salary'] > 50000]
print(result)
```

```python
# Q13. E-Commerce: Find products costing more than Rs. 1000.
prices = [499, 1299, 799, 2499, 999, 1599]
result = [ex for ex in prices if ex > 1000]
print(result)
```

```python
# Q14. E-Commerce Discount: Give a 20% discount only to products costing more than Rs. 2000.
prices = [500, 1200, 2500, 3000, 1500, 5000]
result = [ex * 0.80 for ex in prices if ex > 2000]
print(result)
```

```python
# Q15. Bank Transactions: Positive values represent deposits. Find all deposits and all withdrawals.
transactions = [5000, -2000, 3000, -1000, 7000, -500]
deposits = [ex for ex in transactions if ex > 0]
print(deposits)

withdrawals = [ex for ex in transactions if ex < 0]
print(withdrawals)
```

```python
# Q16. Find names of students who scored 60 or more.
students = [
    {"name": "Amit", "marks": 75},
    {"name": "Rahul", "marks": 45},
    {"name": "Neha", "marks": 90},
    {"name": "Rohit", "marks": 55}
]

result = [s["name"] for s in students if s["marks"] >= 60]
print(result)
```

```python
# Q17. Find names of employees from the IT department.
employees = [
    {"name": "Amit", "department": "IT"},
    {"name": "Rahul", "department": "HR"},
    {"name": "Neha", "department": "IT"},
    {"name": "Rohit", "department": "Sales"}
]

result = [emp["name"] for emp in employees if emp["department"] == "IT"]
print(result)
```

```python
# Q18. Find employees who earn more than Rs. 60,000 and belong to IT.
employees = [
    {"name": "Amit", "department": "IT", "salary": 55000},
    {"name": "Rahul", "department": "HR", "salary": 70000},
    {"name": "Neha", "department": "IT", "salary": 80000},
    {"name": "Rohit", "department": "IT", "salary": 45000}
]

result = [emp["name"] for emp in employees if emp["salary"] > 60000 and emp["department"] == "IT"]
print(result)
```
