# 6. String Indexing, Slicing, Methods & Formatting

```python
data = "Python"
print(data[0]) # indexing
print(data[0:6]) # slicing
print(data[-1]) # - indexing
print(data * 3)
print(*[data] * 3)
```

```python
# len()
data = "Python Programming"
print(len(data))
```

```python
# upper()
data = "Python Programming"
print(data.upper())
```

```python
# lower()
data = "Python Programming"
print(data.lower())
```

```python
# strip()
data = "   Python Programming   "
print(data)
print(data.strip())
```

```python
# lstrip()
data = "   Python Programming   "
print(data)
print(data.lstrip())
```

```python
# rstrip()
data = "   Python Programming   "
print(data)
print(data.rstrip())
```

```python
# replace()
data = "Python Programming"
newData = data.replace("Python", "Java")
print(data)
print(newData)
```

```python
# find()
data = "Python Programming"
print(data.find('P'))
print(data.find('Programming'))
print(data.find('z'))
```

```python
# index()
data = "Python Programming"
print(data.index('P'))
print(data.index('z'))
```

```python
# count()
data = "Python Programming"
print(data.count('P'))
print(data.count('p'))
```

```python
# startswith()
email = "admin@gmail.com"
print(email.startswith('admin'))
```

```python
# endswith()
email = "admin@gmail.com"
print(email.endswith('com'))
print('@' in email)
```

```python
# split()
data = "I Love Python Programming"
newData = data.split()
print(data)
print(newData)
```

```python
# join()
data = ['I', 'Love', 'Python', 'Programming']
newData = " ".join(data)
print(newData)
```

```python
# isalpha()
data = "Invertis"
print(data.isalpha())
```

```python
# isalpha()
data = "Invertis123"
print(data.isalpha())
```

```python
# isdigit()
data = "12345"
print(data.isdigit())
```

```python
# isdigit()
data = "Invertis12345"
print(data.isdigit())
```

```python
# isalnum()
data = "Invertis12345"
print(data.isalnum())
```

```python
# isalnum()
data = "Invertis@12345"
print(data.isalnum())
```

```python
# String Formatting
name = "Sumit"
profile = "HR"
print("User name is: ", name, "and his profile is: ", profile)
```

```python
name = "Sumit"
profile = "HR"
print("User name is: ", profile, "and his profile is: ", name)
```

```python
name = "Sumit"
profile = "HR"
print("User name is:{} and his profile is:{} ".format(name, profile))
```

```python
name = "Sumit"
profile = "HR"
print("User name is:{} and his profile is:{} ".format(profile, name))
```

```python
name = "Sumit"
profile = "HR"
print(f"User name is:{name} and his profile is:{profile}")
```
