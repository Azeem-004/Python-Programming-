# Generators

```python
def even_number(start, end):
    for num in range(start, end + 1):
        if num % 2 == 0:
            return num  # 2

even_number(1, 10)  # 2
```

```python
def even_number(start, end):
    for num in range(start, end + 1):
        if num % 2 == 0:
            yield num  # 2, 4, 6, 8, 10

ex = even_number(1, 10)  # 2, 4, 6, 8, 10
print(next(ex, "Error: iteration Over"))
print(next(ex, "Error: iteration Over"))
print(next(ex, "Error: iteration Over"))
print(next(ex, "Error: iteration Over"))
print(next(ex, "Error: iteration Over"))
print(next(ex, "Error: iteration Over"))  # Error: iteration Over
```

```python
def even_number(start, end):
    for num in range(start, end + 1):
        if num % 2 == 0:
            yield num  # 2, 4, 6, 8, 10

for ex in even_number(1, 10):  # 2, 4, 6, 8, 10
    print(ex)
```

```python
def get_numbers():
    yield 10  # 1
    yield 20  # 2
    yield 30  # 3

for ex in get_numbers():
    print(ex)
```

```python
def get_numbers():  # Waiting
    yield 10  #1
    yield 20  #2
    yield 30  #3 Wait

ex = get_numbers()
print(next(ex))  #10
print(next(ex))  #20
print(type(ex))
```

# Student Records Generator

```python
# Student records
students = [
    {"name": "Aman", "marks": [80, 75, 90]},
    {"name": "Riya", "marks": [35, 40, 30]},
    {"name": "Rahul", "marks": [90, 85, 95]},
    {"name": "Neha", "marks": [60, 70, 65]}
]

def generate_result(student_record):
    for student in student_record:
        name = student["name"]  # Aman
        marks = student["marks"]  # [80, 75, 90]

        avg = sum(marks) / len(marks)
        if avg >= 40:
            status = "Pass"
        else:
            status = "Fail"

        yield {
            "name": name,
            "average": avg,
            "status": status
        }

ex = generate_result(students)
print(next(ex))
print(next(ex))
print(next(ex))
print(next(ex))
```

```python
# Student records
students = [
    {"name": "Aman", "marks": [80, 75, 90]},
    {"name": "Riya", "marks": [35, 40, 30]},
    {"name": "Rahul", "marks": [90, 85, 95]},
    {"name": "Neha", "marks": [60, 70, 65]}
]

def generate_result(student_record):
    for student in student_record:
        name = student["name"]  # Aman
        marks = student["marks"]  # [80, 75, 90]

        avg = sum(marks) / len(marks)
        if avg >= 40:
            status = "Pass"
        else:
            status = "Fail"

        yield {
            "name": name,
            "average": avg,
            "status": status
        }

for result in generate_result(students):
    print(result["name"])
    print(result["average"])
    print(result["status"])
    print("-" * 25)
```
