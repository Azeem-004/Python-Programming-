# 9. Pattern Printing (`for` & `while` Loops)

```python
# Square Star Pattern (for loop)
n = 4
for i in range(1, n + 1):
    for j in range(1, n + 1):
        print("*", end=" ")
    print(end="\n")
```

```python
# Rectangle Star Pattern (for loop)
row = 4
column = 6
for i in range(1, row + 1):
    for j in range(1, column + 1):
        print("*", end=" ")
    print(end="\n")
```

```python
# Right-Angled Triangle Star Pattern (for loop)
n = 5
for i in range(1, n + 1):
    for j in range(1, i + 1):
        print("*", end=" ")
    print(end="\n")
```

```python
# Inverted Right-Angled Triangle Star Pattern (for loop)
n = 5
for i in range(n, 0, -1):
    for j in range(1, i + 1):
        print("*", end=" ")
    print(end="\n")
```

```python
# Number Triangle Pattern - Column-wise 1 to i (for loop)
n = 5
for i in range(1, n + 1):
    for j in range(1, i + 1):
        print(j, end=" ")
    print(end="\n")
```

```python
# Number Triangle Pattern - Row-wise repeating i (for loop)
n = 5
for i in range(1, n + 1):
    for j in range(1, i + 1):
        print(i, end=" ")
    print(end="\n")
```

```python
# Floyd's Triangle Pattern (1 to 15)
n = 5
x = 1
for i in range(1, n + 1): # i = 3
    for j in range(1, i + 1):
        print(x, end=" ")
        x += 1
    print(end="\n")
```

```python
# Alternating 0-1 Triangle Pattern
n = 5
for i in range(1, n + 1):
    for j in range(0, i):
        print((i + j) % 2, end=" ")
    print(end="\n")
```

```python
# Alphabet Triangle Pattern (A to E)
n = 5
for i in range(1, n + 1): # i = 2
    for j in range(0, i): # j = 0, j = 1 , i = 2   Loop 2 time
        print(chr(65 + j), end=" ") # 65 + 1
    print(end="\n")
```

```python
# Diamond Star Pattern (Mirror Pattern)
n = 5
for i in range(1, 2 * n): # i = 7 row = 7 to row 9
    if i <= n: # 7 <= 5
        star = i # 1st row no of star 5
    else:
        star = 2 * n - i # 7th no row star = (2*5) - 7 = 3 star
    space = n - star # space = 5 - 3        space = 2      7th row

    for j in range(0, space): # 2 time
        print("  ", end="")

    for k in range(0, 2 * star - 1): # (2*3) - 1 = 5 >> 5 time
        print("* ", end="")

    print("\n")
```

```python
# Mirror Number Triangle Pattern (1 to 5 and back to 1)
n = 5
for i in range(1, 2 * n): # i = 9   row 9
    if i <= n:
        num = i
    else:
        num = 2 * n - i # num = (2*5) - 9      num = 1
    for j in range(1, num + 1):
        print(j, end=" ")

    print(end="\n")
```

```python
# Reverse Mirror Number Triangle Pattern (num down to 1)
n = 5
for i in range(1, 2 * n):
    if i <= n:
        num = i
    else:
        num = 2 * n - i
    for j in range(num, 0, -1):
        print(j, end=" ")
    print(end="\n")
```

```python
# Square Star Pattern using while loop
n = 4
row = 1

while(row <= n):
    star = 1
    while(star <= n):
        print("* ", end=" ")
        star += 1
    print(end="\n")
    row += 1
```

```python
# Right-Angled Triangle Star Pattern using while loop
num = 5
row = 1

while(row <= num):
    star = 1
    while(star <= row):
        print("* ", end=" ")
        star += 1
    print(end="\n")
    row += 1
```

```python
# Inverted Right-Angled Triangle Star Pattern using while loop
num = 5
row = 1
while(row <= num):
    star = 1
    while(star <= num - row + 1):
        print("* ", end=" ")
        star += 1
    print(end="\n")
    row += 1
```

```python
# Right-Aligned Triangle Star Pattern using while loop
num = 5
row = 1
while(row <= num):
    space = 1
    while(space <= num - row):
        print(" ", end=" ")
        space += 1
    star = 1
    while(star <= row):
        print("*", end=" ")
        star += 1
    print(end="\n")
    row += 1
```

```python
# Inverted Right-Aligned Triangle Star Pattern using while loop
num = 5
row = 1
while(row <= num):
    space = 1
    while(space <= row - 1):
        print(" ", end=" ")
        space += 1
    star = 1
    while(star <= num - row + 1): # 5 - 1 + 1 = 5
        print("*", end=" ")
        star += 1
    print(end="\n")
    row += 1
```

```python
# Pyramid Star Pattern using while loop
num = 5
row = 1
while(row <= num):
    space = 1
    while(space <= num - row): # 5 - 1 = 4
        print(" ", end=" ")
        space += 1
    star = 1
    while(star <= 2 * row - 1): # (2*1) - 1 = 1
        print("*", end=" ")
        star += 1
    print(end="\n")
    row += 1
```
