```python
class EmptyStudent:
    pass

empty_student = EmptyStudent()
print("Object created:", isinstance(empty_student, EmptyStudent))
```

```python
class Student:
    def add_details(self, name, age, course):
        self.name = name
        self.age = age
        self.course = course

    def show_details(self):
        print("Name:", self.name)
        print("Age:", self.age)
        print("Course:", self.course)
        print("-" * 25)

student1 = Student()
student1.add_details("Amit", 23, "Python")
student1.show_details()

student2 = Student()
student2.add_details("Priya", 20, "Java")
student2.show_details()

student3 = Student()
student3.add_details("John", 33, "DBMS")
student3.show_details()
```

```python
class EmployeeRecord:
    def add_employee_details(self, employee_id, name, salary, company):
        self.employee_id = employee_id
        self.name = name
        self.salary = salary
        self.company = company

    def show_employee_details(self):
        print("Employee ID:", self.employee_id)
        print("Employee Name:", self.name)
        print("Employee Salary:", self.salary)
        print("Employee Company:", self.company)
        print("-" * 25)

employee1 = EmployeeRecord()
employee1.add_employee_details(111, "Raj", 12000.0, "HCL")
employee1.show_employee_details()

employee2 = EmployeeRecord()
employee2.add_employee_details(222, "Ansh Chaudhary", 22000.0, "HCL")
employee2.show_employee_details()
```

```python
class Employee:
    def __init__(self, employee_id, name, salary, company="HCL"):
        self.employee_id = employee_id
        self.name = name
        self.salary = salary
        self.company = company

    def show_employee_details(self):
        print("Employee ID:", self.employee_id)
        print("Employee Name:", self.name)
        print("Employee Salary:", self.salary)
        print("Employee Company:", self.company)
        print("-" * 25)

employee1 = Employee(111, "Raj", 12000.0)
employee1.show_employee_details()

employee2 = Employee(222, "Ansh Chaudhary", 22000.0, "HCL")
employee2.show_employee_details()
```

```python
class EmployeeWithCompany:
    COMPANY = "HCL"

    def __init__(self, employee_id, name, salary):
        self.employee_id = employee_id
        self.name = name
        self.salary = salary

    def show_employee_details(self):
        print("Employee ID:", self.employee_id)
        print("Employee Name:", self.name)
        print("Employee Salary:", self.salary)
        print("Employee Company:", EmployeeWithCompany.COMPANY)
        print("-" * 25)

EmployeeWithCompany(111, "Raj", 12000.0).show_employee_details()
EmployeeWithCompany(222, "Ansh Chaudhary", 22000.0).show_employee_details()
```

```python
class BankAccount:
    def __init__(self, account_number=0, account_holder_name="Unknown", amount=0):
        self.account_number = account_number
        self.account_holder_name = account_holder_name
        self.amount = amount

    def show_account_details(self):
        print("Account Number:", self.account_number)
        print("Account Holder Name:", self.account_holder_name)
        print("Current Balance:", self.amount)
        print("=" * 25)

account1 = BankAccount(123123, "Manoj")
account1.show_account_details()

account2 = BankAccount(4633252, "Pawan", 2000)
account2.show_account_details()
```
