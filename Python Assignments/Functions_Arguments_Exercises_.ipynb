```python
# Exercise 1: Student Greeting System
# Question: Create a function that accepts a student's name and course.
# If the course is not given, use "Python Programming".

def greet_student(name, course="Python Programming"):
    print("Hello", name)
    print("Course:", course)

greet_student("Azeem")
greet_student("Azeem", "Data Analytics")
```

```python
# Exercise 2: Shopping Discount Calculator
# Question: Calculate discount based on bill amount.
# Up to ₹999 = 5%, ₹1,000–₹4,999 = 10%, ₹5,000–₹9,999 = 15%, ₹10,000 or more = 20%.
# A custom discount can be given instead. Discount must be from 0% to 100%.

def calculate_discount(bill, discount=None):
    if bill <= 0:
        print("Invalid Bill Amount")
        return

    if discount is None:
        if bill < 1000:
            discount = 5
        elif bill < 5000:
            discount = 10
        elif bill < 10000:
            discount = 15
        else:
            discount = 20

    if discount < 0 or discount > 100:
        print("Invalid Discount")
        return

    discount_amount = bill * discount / 100
    final_amount = bill - discount_amount

    print("Original Bill: ₹", bill)
    print("Discount:", discount, "%")
    print("Discount Amount: ₹", discount_amount)
    print("Final Amount: ₹", final_amount)

calculate_discount(850)
calculate_discount(2500)
calculate_discount(7500)
calculate_discount(12000)
calculate_discount(7500, 7)
```

```python
# Exercise 3: Employee Details
# Question: Create a function that accepts any number of employee details,
# such as department, salary, city and experience.

def show_employee_details(**details):
    for key, value in details.items():
        print(key, ":", value)

show_employee_details(
    department="Data Analytics",
    salary=35000,
    city="Bareilly",
    experience="Fresher"
)
```

```python
# Exercise 4: Student Result
# Question: Accept a student's name, any number of subject marks,
# and extra details like roll number and semester.

def student_result(name, *marks, **details):
    print("Student Name:", name)
    print("Subject Marks:", marks)

    if len(marks) > 0:
        print("Total Marks:", sum(marks))
        print("Average Marks:", sum(marks) / len(marks))
    else:
        print("No marks entered")

    for key, value in details.items():
        print(key, ":", value)

student_result("Azeem", 85, 90, 78, roll_number=101, semester="2nd")
```

```python
# Exercise 5: Shopping Cart and Customer Details
# Question: Pass prices from a list and customer details from a dictionary
# to functions using argument unpacking.

def calculate_total(*prices):
    print("Cart Total: ₹", sum(prices))

def show_customer(name, city, **details):
    print("Customer Name:", name)
    print("City:", city)
    for key, value in details.items():
        print(key, ":", value)

prices = [299, 499, 199]
customer = {
    "name": "Azeem",
    "city": "Bareilly",
    "membership": "Gold"
}

calculate_total(*prices)
show_customer(**customer)
```

```python
# Exercise 6: Average Marks Calculator
# Question: Calculate the average of marks. Reject empty input and invalid marks.

def calculate_average(*marks):
    if len(marks) == 0:
        print("No marks entered")
        return

    for mark in marks:
        if not isinstance(mark, (int, float)) or mark < 0 or mark > 100:
            print("Invalid marks")
            return

    average = sum(marks) / len(marks)
    print("Average Marks:", average)

calculate_average(80, 90, 75)
calculate_average()
calculate_average(80, 120, 75)
```
