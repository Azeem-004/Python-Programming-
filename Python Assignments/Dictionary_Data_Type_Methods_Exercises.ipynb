# 8. Dictionary Data Type, Methods & Exercises

```python
data = {}
print(type(data))
print(data)
```

```python
data = dict()
print(type(data))
print(data)
```

```python
# Values can be duplicate
data = {"name1": "Mohit", "age": 23, "Att": True, "name2": "Mohit"}
print(data)
```

```python
# Keys can't be duplicate
data = {"name": "Mohit", "age": 23, "Att": True, "name": "Amit"}
print(data)
```

```python
# Dict are Mutable
data = {"name": "Mohit", "age": 23, "Att": True}
print(data)

data["name"] = "Priya"
print(data)

print(data["name"]) # Using Key >>
# print(data["college"]) # Using Key >>

print(data.get("name"))
print(data.get("college"))
```

```python
# len()
data = {"name": "Mohit", "age": 23, "Att": True}
print(len(data))

# values()
print(data.values())

# items()
print(data.items())
```

```python
# pop()
data = {"name": "Mohit", "age": 23, "Att": True}
data.pop("age")
print(data)
```

```python
# pop()
data = {"name": "Mohit", "age": 23, "Att": True}
data.pop("age")
```

```python
# popitem()
data = {"name": "Mohit", "age": 23, "Att": True}
data.popitem()
```

```python
# clear()
data = {"name": "Mohit", "age": 23, "Att": True}
data.clear()
print(data)
```

```python
# Reference Assignment vs copy()
data = {"name": "Mohit", "age": 23, "Att": True}
newdata = data
# print(newdata)

data["name"] = "Rohit"
print(newdata)
print(data)
```

```python
# copy()
data = {"name": "Mohit", "age": 23, "Att": True}
newdata = data.copy()
# print(newdata)

data["name"] = "Rohit"
print(newdata)
print(data)
```

```python
data = {"name": "Mohit", "age": 23, "Att": True}
print("name" in data)
print("Mohit" in data.values())
```

```python
data = {"name": "Mohit", "age": 23, "Att": True}
print("name" not in data)
print("Mohit" not in data.values())
```

## Dictionary Exercises (1 to 10)

```python
# 1. Check Student Result
# Question:
# Given a dictionary containing student marks, use a loop to print whether each student has Passed or Failed.
# Condition:
# Marks >= 40 -> Pass
# Marks < 40 -> Fail

students = {
    "Rahul": 75,
    "Amit": 32,
    "Priya": 88,
    "Neha": 39,
    "Rohit": 55
}

for name, marks in students.items():
    if marks >= 40:
        print(name, " Pass")
    else:
        print(name, " Fail")
```

```python
# 2. Student Performance Using Nested If
# Question:
# Print the performance category of each student.
# Conditions:
# Marks >= 80 -> Excellent
# Marks >= 60 -> Good
# Marks >= 40 -> Average
# Marks < 40 -> Fail

students = {
    "Rahul": 85,
    "Amit": 35,
    "Priya": 72,
    "Neha": 48
}

for name, Marks in students.items():
    if(Marks > 40):
        if(Marks >= 80):
            print(name, " Excellent")
        elif(Marks >= 60):
            print(name, " Good")
        elif(Marks >= 40):
            print(name, " Average")
    else:
        print(name, " Fail")
```

```python
# 3. Product Stock Management
# Question:
# A dictionary contains products and their available stock. Use a loop and nested conditions to display the stock status.
# Conditions:
# Stock = 0 -> Out of Stock
# Stock <= 5 -> Low Stock
# Stock > 5 -> Available

products = {
    "Laptop": 10,
    "Mouse": 3,
    "Keyboard": 0,
    "Monitor": 7,
    "Printer": 2
}

for product, stock in products.items():
    if stock > 0:
        if stock > 5:
            print(product, ": Available")
        else:
            print(product, ": Low Stock")
    else:
        print(product, ": Out of Stock")
```

```python
# 4. Employee Salary and Department
# Question:
# The following dictionary contains employee salary and department. Display the employee category.
# Conditions:
# Salary >= 50,000:
#   IT -> Senior IT Employee
#   HR -> Senior HR Employee
#   Other department -> Senior Employee
# Salary < 50,000 -> Junior Employee

employees = {
    "Rahul": {"salary": 60000, "department": "IT"},
    "Amit": {"salary": 40000, "department": "HR"},
    "Priya": {"salary": 70000, "department": "IT"},
    "Neha": {"salary": 45000, "department": "Finance"}
}

for name, data in employees.items():
    salary = data["salary"] # salary = 60000
    department = data["department"]

    if salary >= 50000:
        if department == "IT":
            print(name, " Senior IT Employee")
        elif department == "HR":
            print(name, " Senior HR Employee")
        else:
            print(name, " Senior Employee")
    else:
        print(name, " Junior Employee")
```

```python
# 5. Check Even/Odd Values
# Question:
# Given a dictionary, check whether each value is even or odd.
# If the value is:
# Even and greater than 20 -> Large Even
# Even and <= 20 -> Even
# Odd -> Odd

data = {
    "a": 10,
    "b": 15,
    "c": 25,
    "d": 30,
    "e": 40
}

for key, val in data.items():
    if val % 2 == 0:
        if val > 20:
            print(key, ":", val, "-> Large Even")
        else:
            print(key, ":", val, "-> Even")
    else:
        print(key, ":", val, "-> Odd")
```

```python
# 6. Shopping Cart Classification
# Question:
# A dictionary contains products and their prices.
# Conditions:
# Price >= 1000:
#   Price >= 5000 -> Premium Product
#   Otherwise -> Expensive Product
# Price < 1000 -> Affordable Product

products = {
    "Mobile": 15000,
    "Mouse": 500,
    "Laptop": 60000,
    "Keyboard": 1200
}

for item, price in products.items():
    if price >= 1000:
        if price >= 5000:
            print(item, ": Premium Product")
        else:
            print(item, ": Expensive Product")
    else:
        print(item, ": Affordable Product")
```

```python
# 7. Student Eligibility
# Question:
# The dictionary contains student attendance and marks.
# Conditions:
# Attendance >= 75:
#   Marks >= 40 -> Eligible
#   Marks < 40 -> Failed in Exam
# Attendance < 75 -> Not Eligible

students = {
    "Rahul": {"attendance": 85, "marks": 70},
    "Amit": {"attendance": 60, "marks": 80},
    "Priya": {"attendance": 90, "marks": 35},
    "Neha": {"attendance": 78, "marks": 55}
}

for name, info in students.items():
    if info["attendance"] >= 75:
        if info["marks"] >= 40:
            print(name, ": Eligible")
        else:
            print(name, ": Failed in Exam")
    else:
        print(name, ": Not Eligible")
```

```python
# 8. Bank Account Classification
# Question:
# A dictionary contains customer names and account balances.
# Conditions:
# Balance > 0:
#   Balance >= 10,000 -> Premium Account
#   Otherwise -> Regular Account
# Balance = 0 -> No Balance
# Balance < 0 -> Overdraft

accounts = {
    "Rahul": 25000,
    "Amit": 5000,
    "Priya": 0,
    "Neha": -2000
}

for customer, balance in accounts.items():
    if balance > 0:
        if balance >= 10000:
            print(customer, ": Premium Account")
        else:
            print(customer, ": Regular Account")
    elif balance == 0:
        print(customer, ": No Balance")
    else:
        print(customer, ": Overdraft")
```

```python
# 9. Nested Dictionary - Employee Experience
# Question:
# Classify employees based on salary and experience.
# Conditions:
# Salary >= 50,000:
#   Experience >= 5 -> Senior
#   Otherwise -> Mid-Level
# Salary < 50,000:
#   Experience >= 3 -> Experienced Junior
#   Otherwise -> Fresher/Junior

employees = {
    "Rahul": {"salary": 65000, "experience": 6},
    "Amit": {"salary": 55000, "experience": 3},
    "Priya": {"salary": 40000, "experience": 4},
    "Neha": {"salary": 30000, "experience": 1}
}

for name, info in employees.items():
    if info["salary"] >= 50000:
        if info["experience"] >= 5:
            print(name, ": Senior")
        else:
            print(name, ": Mid-Level")
    else:
        if info["experience"] >= 3:
            print(name, ": Experienced Junior")
        else:
            print(name, ": Fresher/Junior")
```

```python
# Question 10:
# The following dictionary contains students' marks and attendance.
# Conditions:
# If attendance < 75 -> Not Eligible
# If attendance >= 75:
#   Marks >= 80 -> Distinction
#   Marks >= 60 -> First Division
#   Marks >= 40 -> Pass
#   Marks < 40 -> Fail

students = {
    "Rahul": {"marks": 85, "attendance": 90},
    "Amit": {"marks": 35, "attendance": 80},
    "Priya": {"marks": 65, "attendance": 70},
    "Neha": {"marks": 45, "attendance": 85},
    "Rohit": {"marks": 90, "attendance": 60}
}

for name, info in students.items():
    if info["attendance"] < 75:
        print(name, "Not Eligible")
    else:
        if info["marks"] >= 80:
            print(name, "Distinction")
        elif info["marks"] >= 60:
            print(name, "First Division")
        elif info["marks"] >= 40:
            print(name, "Pass")
        else:
            print(name, "Fail")
```
