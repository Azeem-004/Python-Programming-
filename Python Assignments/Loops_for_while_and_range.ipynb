```python
range(10) # Ending
range(1, 10) # Start, End
range(1, 10, 1) # Start, End, Step
```

```python
for i in range(1, 10, 1):
    print(i)
```

```python
# Iterating over a List
data = [43, 54, 65, 23, 76, 87]
for i in data:
    print(i)
print(data)
```

```python
# Iterating over a Tuple
data = (43, 54, 65, 23, 76, 87)
for i in data:
    print(i)
```

```python
# Reverse Loop using range()
for i in range(10, 0, -1):
    print(i)
```

```python
# Iterating over a String
data = "PYTHON"
for i in data:
    print(i)
```

```python
# Print Even Numbers from 1 to 20 using if condition
for i in range(1, 21, 1):
    if(i % 2 == 0):
        print(i)
```

```python
# Print Even Numbers from 2 to 20 using step size 2
for i in range(2, 21, 2):
    print(i)
```

```python
# Print Odd Numbers from 1 to 20
for i in range(1, 21, 1):
    if(i % 2 != 0):
        print(i)
```

```python
# Print Multiples of 5 from 5 to 50
for i in range(5, 51, 1):
    if i % 5 == 0:
        print(i)
```

```python
# Sum of Numbers from 1 to 10
sum = 0
for i in range(1, 11, 1):
    sum += i # sum = sum + i   sum = 0 + 1   sum = 1
print(sum)
```

```python
# Sum of Even Numbers from 1 to 20
sum = 0
for i in range(1, 21, 1):
    if(i % 2 == 0):
        sum += i
print(sum)
```

```python
# Count Even and Odd Numbers from 1 to 10
countEven = 0
countOdd = 0
for i in range(1, 11, 1):
    if i % 2 == 0:
        countEven += 1
    else:
        countOdd += 1
print(countEven)
print(countOdd)
```

```python
# Count Positive and Negative Numbers in a List using while loop
data = [22, 32, 45, -2, 87, -1, 89, -4]
posCount = 0
negCount = 0
i = 0
while(i < len(data)):
    if data[i] > 0:
        posCount += 1
    else:
        negCount += 1
    i += 1
print(posCount)
print(negCount)
```

```python
# Count occurrences of 'P' or 'p' in a String using while loop
data = "Python Programming p"
count = 0
i = 0
while(i < len(data)):
    if data[i] == "P" or data[i] == 'p':
        count += 1
    i += 1
print(count)
```
