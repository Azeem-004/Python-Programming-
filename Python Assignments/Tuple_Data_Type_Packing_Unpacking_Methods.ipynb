# 2. Tuple Data Type, Packing, Unpacking & Methods

```python
# Tuple
# Important Point: Parentheses () are commonly used to create tuples
data = (10, 20, 30, 40, 50)
print(type(data))
```

```python
# But parentheses are not actually mandatory
# The comma , is what makes a tuple, not the parentheses.
data2 = 10, 20, 30, 40, 50
print(type(data2))
```

```python
data3 = (10)
print(type(data3))
# Output: <class 'int'>

# Why?
# Because (10) is not a tuple. It is simply an integer 10 enclosed in parentheses.

# Python treats it as:
# data3 = 10
```

```python
# How to create a single-element tuple?
# You must use a comma:
data4 = (10,)
print(type(data4))

# Remember this rule:
# Code          Type
# data = (10)   int
# data = (10,)  tuple
# data = 10,    tuple
# data = (10, 20) tuple
```

```python
# Tuple Indexing
# Tuples are ordered and immutable.
data = (10, 20, 30, 40, 50)
print(data[2]) # Indexing is used to access individual elements of a tuple and Tuple indexing starts from 0.
print(data[-1]) # Negative indexing starts from -1 from the right side.
```

```python
# A tuple allows duplicate elements.
data = (10, 20, 30, 40, 50, 30) # Duplicate Elements
print(data)
```

```python
# Tuple is Immutable
# A tuple is immutable, which means its elements cannot be changed, updated, or deleted after creation.
data = (10, 20, 30, 40, 50)
data[0] = 88 # tries to change the first element from 10 to 88. Therefore, Python raises a TypeError.

# TypeError: 'tuple' object does not support item assignment
```

```python
# Tuple Packing:
studentData = "Aman", "MCA AI", 78.65 # Tuple Packing: Tuple Packing means storing multiple values into a single tuple without explicitly using parentheses.
print(type(studentData))

# Tuple Unpacking
name, course, avgMarks = studentData # Tuple Unpacking means extracting the elements of a tuple into separate variables.
print(name)
# Note: The number of variables should normally match the number of tuple elements.
```

```python
record = "Amit", "MCA", 23, True # Tuple Packing
print(record[0])
print(record[1])
print(record)

name, course, age, att = record
# name, course, age, att = "Amit", "MCA", 23, True    Tuple Unpacking

print(name)

# Unpacking Using *
# * in tuple unpacking is called extended unpacking. It allows one variable to collect multiple remaining values.
n, *ab = record
# Note: The * operator collects the remaining elements into a list.
print(n)
print(ab)
```

```python
record = "Amit", "MCA", 23, True # Tuple Packing
name, course, age = record # ValueError: too many values to unpack (expected 3)
# Note: The number of variables should normally match the number of tuple elements.
```

```python
record = "Amit", "MCA", 23, True # Tuple Packing
name, *ab, att = record
print(name)
print(ab)
print(att)
```

```python
# len() Returns the total number of elements
data = (10, 20, 30, 40, 50, 30)
print(len(data))

# count() : Counts how many times a value occurs
print(data.count(30))

print(sum(data)) # Returns the sum of numeric elements
print(max(data)) # Returns the largest element
print(min(data)) # Returns the smallest element
print(data.index(10)) # Returns the index of the first occurrence
```
